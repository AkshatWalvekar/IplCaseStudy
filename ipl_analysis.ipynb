```python
import numpy as np

data = np.genfromtxt(
    "deliveries.csv",
    delimiter=",",
    dtype=str,
    skip_header=1
)

print("Shape:", data.shape)
```

```python
match_id = data[:, 0].astype(int)
batting_team = data[:, 2]
over = data[:, 4].astype(int)
batter = data[:, 6]
bowler = data[:, 7]
batsman_runs = data[:, 9].astype(int)

print("Columns extracted ")
```

```python
print(match_id[:5])
print(batter[:5])
print(batsman_runs[:5])
print(over[:5])
```

```python
# Task 1: Total Runs per Match

unique_matches = np.unique(match_id)

total_runs_per_match = np.array([
    (m, np.sum(batsman_runs[match_id == m]))
    for m in unique_matches
])

# Show first 5 results
total_runs_per_match[:5]
```

```python
# Task 2: Top 5 Batters

players = np.unique(batter)

# Total runs per player
runs_per_player = np.array([
    np.sum(batsman_runs[batter == p])
    for p in players
])

# Get indices of top 5 players (descending order)
top_indices = np.argsort(runs_per_player)[::-1][:5]

# Final result
top_5_batters = list(zip(players[top_indices], runs_per_player[top_indices]))

top_5_batters
```
