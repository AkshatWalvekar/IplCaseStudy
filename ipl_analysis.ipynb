```python
import numpy as np

data = np.genfromtxt(
    "deliveries.csv",
    delimiter=",",
    dtype=str,
    skip_header=1
)

print("Shape:", data.shape)
```

```python
match_id = data[:, 0].astype(int)
batting_team = data[:, 2]
over = data[:, 4].astype(int)
batter = data[:, 6]
bowler = data[:, 7]
batsman_runs = data[:, 9].astype(int)

print("Columns extracted ")
```

```python
print(match_id[:5])
print(batter[:5])
print(batsman_runs[:5])
print(over[:5])
```

```python
# Task 1: Total Runs per Match

unique_matches = np.unique(match_id)

total_runs_per_match = np.array([
    (m, np.sum(batsman_runs[match_id == m]))
    for m in unique_matches
])

# Show first 5 results
total_runs_per_match[:5]
```

```python
# Task 2: Top 5 Batters

players = np.unique(batter)

# Total runs per player
runs_per_player = np.array([
    np.sum(batsman_runs[batter == p])
    for p in players
])

# Get indices of top 5 players (descending order)
top_indices = np.argsort(runs_per_player)[::-1][:5]

# Final result
top_5_batters = list(zip(players[top_indices], runs_per_player[top_indices]))

top_5_batters
```

```python
# Task 3: Strike Rate

players = np.unique(batter)

# Total runs per player
total_runs = np.array([
    np.sum(batsman_runs[batter == p])
    for p in players
])

# Balls faced per player
balls_faced = np.array([
    np.sum(batter == p)
    for p in players
])

# Strike rate calculation
strike_rates = (total_runs / balls_faced) * 100

# Combine results
strike_rate_result = list(zip(players, strike_rates))

# Show first 5
strike_rate_result[:5]
```

```python
# Task 4: Economy Rate of Bowlers

bowlers = np.unique(bowler)

# Total runs conceded per bowler
runs_conceded = np.array([
    np.sum(batsman_runs[bowler == b])
    for b in bowlers
])

# Total balls bowled per bowler
balls_bowled = np.array([
    np.sum(bowler == b)
    for b in bowlers
])

# Convert balls to overs
overs = balls_bowled / 6

# Economy rate
economy_rate = runs_conceded / overs

# Combine results
economy_result = list(zip(bowlers, economy_rate))

# Show first 5
economy_result[:5]
```

```python
# Task 5: Runs per Over (1 to 20)

runs_per_over = np.array([
    np.mean(batsman_runs[over == i])
    for i in range(1, 21)
])

runs_per_over
```

```python
# Task 6: Boundary Analysis

fours = np.sum(batsman_runs == 4)
sixes = np.sum(batsman_runs == 6)

print("Total Fours:", fours)
print("Total Sixes:", sixes)
```

```python
# Task 7: Death Overs Analysis

# Mask for overs 16–20
death_mask = (over >= 16) & (over <= 20)

# Total runs in death overs
death_runs = np.sum(batsman_runs[death_mask])

death_runs
```

```python
teams = np.unique(batting_team)

death_runs_team = np.array([
    np.sum(batsman_runs[(batting_team == t) & (over >= 16) & (over <= 20)])
    for t in teams
])

top_team = teams[np.argmax(death_runs_team)]

top_team
```

```python
# Task 8: Highest Scoring Match

unique_matches = np.unique(match_id)

# Total runs per match
total_runs = np.array([
    np.sum(batsman_runs[match_id == m])
    for m in unique_matches
])

# Find max
idx = np.argmax(total_runs)

highest_match = unique_matches[idx]
highest_score = total_runs[idx]

highest_match, highest_score
```
